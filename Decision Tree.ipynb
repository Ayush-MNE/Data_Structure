1. What is a Decision Tree, and how does it work	

Ans. A **Decision Tree** is a supervised learning algorithm used for both **classification** and **regression** tasks. It models decisions using a tree-like structure, where each internal node represents a decision based on a feature, each branch represents an outcome, and each leaf node represents a final prediction.

### **How It Works**
1. **Splitting**: The dataset is split based on a chosen feature that best separates the data into distinct classes (classification) or predicts a target variable (regression).
2. **Selection of Splitting Criteria**:
   - For **classification**, the tree uses measures like **Gini Impurity** or **Entropy (Information Gain)** to decide the best split.
   - For **regression**, it minimizes **Mean Squared Error (MSE)** or **Mean Absolute Error (MAE)**.
3. **Recursive Splitting**: The process repeats for each child node until one of the stopping conditions is met (e.g., maximum depth, minimum samples per leaf).
4. **Pruning (Optional)**: To avoid overfitting, branches with little predictive power can be removed through **pre-pruning** (early stopping) or **post-pruning** (removing nodes after training).

### **Advantages**
- **Easy to interpret** and visualize.
- **Handles both numerical and categorical data**.
- **Does not require feature scaling**.
- **Captures non-linear relationships**.

### **Disadvantages**
- **Prone to overfitting**, especially with deep trees.
- **Sensitive to small variations in data**, leading to different tree structures.
- **Less accurate than ensemble methods** like Random Forest or Gradient Boosting.


2.  What are impurity measures in Decision Trees ?

Ans. **Impurity measures** in Decision Trees determine how "mixed" a node is. Lower impurity means the node contains mostly one class, making it a better split. The three main impurity measures are:

### **1. Gini Impurity (Used in CART)**
Measures the probability of incorrectly classifying a randomly chosen element if it were labeled according to the node’s class distribution.

- Ranges from **0 (pure)** to **0.5 (maximum impurity for two classes)**.
- Faster to compute than entropy.

### **2. Entropy (Used in ID3, C4.5)**
Measures the disorder in a node. Based on **Information Gain**, which quantifies how much uncertainty is reduced after a split.
- Ranges from **0 (pure node)** to **log₂(c) (maximum impurity with c classes)**.
- Slower than Gini but more informative for smaller datasets.

### **3. Variance Reduction (For Regression Trees)**
Measures the reduction in variance when splitting a node.

- The split that minimizes the variance of the target variable is chosen.
- Helps in **regression tasks** where the goal is to reduce prediction error.



3. What is the mathematical formula for Gini Impurity?

Ans. The **mathematical formula** for **Gini Impurity** is:
Gini = 1 - (sum_ (p_i^2))

where:  
- \( c \) = number of classes,  
- \( p_i \) = proportion of class \( i \) in the node.

### **Explanation**:
- Gini measures the probability of **misclassifying** a randomly chosen element.
- A **pure** node (only one class) has **Gini = 0**.
- A **completely mixed** node (equal distribution of classes) has **higher Gini**.

### **Example Calculation**:
If a node has **100 samples**, with **60 from Class A** and **40 from Class B**:

p_A = \frac{60}/{100} = 0.6,  p_B = frac{40}/{100} = 0.4

Gini = 1 - (0.6^2 + 0.4^2) = 1 - (0.36 + 0.16) = 1 - 0.52 = 0.48

4.  What is the mathematical formula for Entropy ?	

Ans. The **mathematical formula** for **Entropy** in a Decision Tree is:

\[
Entropy = - \sum_( p_i * log_2 p_i)
where:  
- \( c \) = number of classes,  
- ( p_i ) = proportion of class ( i ) in the node.

### **Explanation**:
- **Entropy measures the "disorder" or uncertainty** in a node.

- If classes are **equally distributed**, **Entropy is maximum**.

### **Example Calculation**:
For a node with **100 samples**, where **60 belong to Class A** and **40 to Class B**:
p_A = \frac{60}/{100} = 0.6,  p_B = \frac{40}/{100} = 0.4
Entropy = - (0.6 \log_2 0.6 + 0.4 \log_2 0.4)

Approximating values:
\log_2 0.6 \approx -0.737, \quad \log_2 0.4 \approx -1.322
Entropy = - (0.6 * -0.737 + 0.4 * -1.322)
        = - (-0.4422 - 0.5288) = 0.971

5. What is Information Gain, and how is it used in Decision Trees	

Ans. ### **What is Information Gain?**  
**Information Gain (IG)** measures how much **uncertainty (entropy) is reduced** after splitting a dataset on a feature. It helps a Decision Tree decide which feature to split on at each step.

### **Mathematical Formula**
\[
IG = Entropy(Parent) - sum_(frac{N_j}/{N} * Entropy(Child_j))
\]
where:  
- **Entropy(Parent)** = Entropy before splitting.  
- **k** = Number of child nodes (usually 2 for binary trees).  
- **N_j** = Number of samples in child node \( j \).  
- **N** = Total samples in the parent node.  
- **Entropy(Child_j)** = Entropy of child node \( j \).  

### **How It Works in Decision Trees**
1. **Compute the entropy** of the entire dataset (before splitting).
2. **Split the dataset** using a feature.
3. **Compute entropy of child nodes** and take a weighted sum.
4. **Subtract this from the original entropy** to get Information Gain.
5. **Choose the feature with the highest IG** as the best splitting feature.

### **Example Calculation**
If the parent node has **Entropy = 0.971** and splits into two child nodes with entropies **0.811 (60 samples)** and **0.722 (40 samples)**:


IG = 0.971 - \left( \frac{60}/{100} \times 0.811 + \frac{40}/{100} \times 0.722 \right)

IG = 0.971 - (0.4866 + 0.2888) = 0.971 - 0.7754 = 0.1956
### **Why is Information Gain Important?**
- Helps **select the best feature** for splitting at each step.
- A **higher IG means better feature selection**, leading to an **optimal tree**.
- **Minimizes uncertainty**, making the tree more effective.

### **1. Definition**
- **Gini Impurity**: Measures the probability of incorrectly classifying a randomly chosen element if labeled according to the class distribution.
- **Entropy**: Measures the uncertainty (or disorder) in the dataset based on information theory.

### **2. Formula**
- **Gini Impurity**:  
  \[
  Gini = 1 - \sum p_i^2
  \]
  where \( p_i \) is the probability of class \( i \).

  \[

### **5. Computational Efficiency**
- **Gini is faster** to compute because it doesn’t involve logarithms.
- **Entropy is slower** as it requires calculating logarithms.

### **6. Preference in Decision Trees**
- **Gini (default in CART trees like scikit-learn)** is preferred for speed and simplicity.
- **Entropy (used in ID3, C4.5, and some other algorithms)** is preferred when a probabilistic approach (information gain) is needed.

- **Gini (Gini Gain)**: Measures the reduction in impurity.

- **Entropy might perform better with more balanced classes**

7.  What is the mathematical explanation behind Decision Trees

Ans. ### **Mathematical Explanation Behind Decision Trees (Without Formulas)**

A **Decision Tree** is a predictive model that splits data into smaller groups based on certain conditions. The goal is to create groups that are as pure (homogeneous) as possible. This process is guided by mathematical principles that determine the best way to split the data.

---

### **1. Splitting the Data**
At each step, the algorithm looks for the best way to divide the dataset into two parts. It does this by selecting a feature and a threshold value that result in the best separation. The quality of a split is determined by how much it reduces the impurity in the resulting groups.

For **classification problems**, impurity refers to how mixed the different classes (categories) are in a group. A pure group contains only one class, while an impure group has a mix of multiple classes.

For **regression problems**, impurity refers to how spread out the numerical values are within a group. The goal is to make each group as close as possible in value.

---

### **2. Measuring Impurity**
To decide the best split, the algorithm calculates impurity using certain measures:

- **For classification**, impurity is measured by checking how evenly distributed different classes are within a group. A group with a mix of multiple classes is considered highly impure, while a group with only one class is completely pure.
- **For regression**, impurity is measured by checking how far the values deviate from the average. A group where all values are close together is more pure than a group with a wide range of values.

The best split is the one that reduces impurity the most.

---

### **3. Stopping the Tree Growth**
The tree continues splitting until certain conditions are met. These stopping conditions include:
- Reaching a predefined depth (number of levels in the tree).
- Having too few samples left in a group to make a meaningful split.
- No further splits significantly improving the separation.

Stopping early helps prevent the tree from growing too large and overfitting the training data.

---

### **4. Preventing Overfitting (Pruning)**
A tree that is too deep may fit the training data perfectly but fail to generalize to new data. To avoid this, pruning techniques are used:

- **Pre-pruning:** Stops the tree from growing when further splits don’t add much value.
- **Post-pruning:** First builds a large tree and then removes unnecessary splits that don’t improve accuracy significantly.

By pruning the tree, we strike a balance between capturing patterns in the data and keeping the model simple.

---

### **5. Recursive Tree Building**
Decision Trees use a **greedy approach**, meaning they make the best decision at each step without looking ahead. The process follows these steps:

1. Start with all the data at the root.
2. Find the best feature and threshold to split the data.
3. Divide the data into two groups based on the split.
4. Repeat the process for each new group until a stopping condition is met.

This results in a tree-like structure where each split represents a decision.

---

### **Summary**
- Decision Trees work by **splitting data into smaller, purer groups**.
- They use **impurity measures** to determine the best splits.
- The tree grows **until stopping conditions** are met.
- **Pruning** helps prevent overfitting.
- The tree-building process follows a **recursive, greedy approach**.



8. What is Pre-Pruning in Decision Trees

Ans. ### **Pre-Pruning in Decision Trees**

**Pre-pruning (Early Stopping)** is a technique used to prevent a Decision Tree from growing too deep and overfitting the training data. Instead of allowing the tree to grow until it perfectly classifies the training data, pre-pruning **stops the tree’s growth early** based on certain conditions.

---

### **How Pre-Pruning Works**
During the tree-building process, the algorithm evaluates whether to continue splitting a node based on predefined criteria. If any of these conditions are met, the tree stops splitting at that node, making it a **leaf node** instead of growing further.

---

### **Common Stopping Criteria for Pre-Pruning**
1. **Maximum Depth** – Limits the number of levels in the tree. Once the tree reaches this depth, it stops further splits.
2. **Minimum Samples per Split** – Requires a minimum number of samples in a node before a split is allowed. If the node has too few samples, it becomes a leaf.
3. **Minimum Samples per Leaf** – Ensures that each leaf has at least a certain number of samples, preventing overly small, specialized leaves.
4. **Minimum Information Gain (or Gini Reduction)** – Stops splitting if the reduction in impurity (Gini or Entropy) is too small, meaning further splits don’t significantly improve the model.
5. **Chi-Square Test (for categorical features)** – Checks whether a split significantly improves classification. If not, the split is rejected.

---

### **Advantages of Pre-Pruning**
- **Prevents Overfitting** – Stops the tree from memorizing the training data too much.
- **Improves Generalization** – Makes the tree perform better on new, unseen data.
- **Reduces Computational Cost** – Stops unnecessary splits, making the model faster to train and use.
- **Simplifies the Model** – Produces a smaller, more interpretable tree.

---

### **Limitations of Pre-Pruning**
- **Risk of Underfitting** – Stopping the tree too early may prevent it from capturing important patterns in the data.
- **Choosing the Right Criteria is Difficult** – Setting thresholds (like max depth or min samples) requires experimentation and validation.

---

### **Pre-Pruning vs. Post-Pruning**
| Feature          | Pre-Pruning (Early Stopping) | Post-Pruning (Prune After Full Growth) |
|-----------------|----------------------------|----------------------------------------|
| **When Applied**  | During tree growth        | After full tree is built             |
| **How It Works**  | Stops splitting early based on conditions | Removes unnecessary branches from a fully grown tree |
| **Risk**        | May stop too soon (underfitting) | May require extra computation but prevents overfitting |
| **Computation**  | Faster, since it avoids unnecessary splits | Slower, as it builds the full tree first |

---

### **Example of Pre-Pruning in Practice**
Imagine we are building a decision tree to classify loan applicants as **low-risk or high-risk**. If we set a **maximum depth of 5**, the tree will never grow beyond 5 levels, even if deeper splits might improve accuracy. This helps keep the model simple and prevents overfitting.

9. What is Post-Pruning in Decision Trees

Ans. ### **Post-Pruning in Decision Trees**  

**Post-pruning** is a technique used to improve a Decision Tree’s generalization ability by removing unnecessary branches after the tree has been fully grown. Unlike **pre-pruning**, which stops the tree from growing beyond a certain point, post-pruning **first allows the tree to grow fully** and then trims back parts that do not significantly improve accuracy.  

---

### **How Post-Pruning Works**
1. **Build the Full Tree** – The Decision Tree is grown until it perfectly fits (or nearly fits) the training data, meaning it might be overfitting.  
2. **Evaluate Each Node** – The tree is examined to see if removing certain branches (subtrees) improves overall performance.  
3. **Remove Weak Branches** – Nodes or branches that contribute little to prediction accuracy are removed, replacing them with leaf nodes.  
4. **Validate on a Separate Dataset** – A validation dataset is used to check if pruning improves generalization (reduces overfitting).  

---

### **Methods of Post-Pruning**
1. **Cost Complexity Pruning (CCP)**
   - Assigns a cost to having more branches and removes nodes that don’t provide enough benefit.
   - Balances tree complexity and accuracy using a **regularization parameter**.
   
2. **Reduced Error Pruning**
   - Tests whether removing a subtree increases or decreases accuracy on validation data.
   - If accuracy stays the same or improves, the subtree is pruned.

3. **Minimum Error Pruning**
   - Uses statistical tests (like chi-square) to check if further splits are significant.
   - If a split does not significantly reduce classification error, it is pruned.

---

### **Advantages of Post-Pruning**
- **Prevents Overfitting** – By removing unnecessary branches, the tree generalizes better to unseen data.  
- **Better Accuracy on Test Data** – A pruned tree often performs better than an unpruned one on real-world data.  
- **More Interpretability** – The tree becomes smaller and easier to understand.  

---

### **Limitations of Post-Pruning**
- **Computationally Expensive** – Requires growing the full tree first, which can be slow for large datasets.  
- **May Prune Too Much** – If not carefully tuned, important splits may be removed, leading to underfitting.  

---

### **Post-Pruning vs. Pre-Pruning**
| Feature          | Pre-Pruning (Early Stopping) | Post-Pruning (After Full Growth) |
|-----------------|----------------------------|----------------------------------------|
| **When Applied**  | During tree growth        | After full tree is built             |
| **How It Works**  | Stops splitting early based on conditions | Removes unnecessary branches from a fully grown tree |
| **Risk**        | May stop too soon (underfitting) | May remove useful branches (if not tuned well) |
| **Computation**  | Faster, avoids unnecessary splits | Slower, as it builds the full tree first |

---

### **Example of Post-Pruning**
Imagine a **decision tree predicting customer churn** in a telecom company. Initially, the tree grows deep and perfectly fits the training data, including small details that may not generalize well. After post-pruning:  
- The tree removes branches that do not significantly impact predictions.  
- The final model is smaller, interpretable, and better at handling new customer data.

11. What is a Decision Tree Regressor

Ans. ### **Decision Tree Regressor: An Overview**  

A **Decision Tree Regressor** is a type of decision tree model used for **predicting continuous numerical values** instead of categorical labels. It works by recursively splitting the data into smaller and smaller subsets, with the goal of minimizing the variance within each group.  

---

### **How a Decision Tree Regressor Works**  

1. **Select the Best Feature to Split**  
   - The algorithm evaluates different features and selects the one that results in the most **homogeneous** (least variable) groups.  

2. **Find the Optimal Split Point**  
   - Instead of classifying data, it determines a threshold that minimizes the variation (spread) of the target variable within each group.  

3. **Recursive Splitting**  
   - The dataset is split repeatedly until a stopping criterion is met, such as **maximum depth, minimum samples per split, or minimum variance reduction**.  

4. **Make Predictions**  
   - At each leaf node, the model predicts a value by **taking the average of the target values** within that node.  

---

### **Mathematical Concept (Without Formulas)**  

- Instead of using classification metrics like **Gini impurity** or **Entropy**, a Decision Tree Regressor **measures impurity based on variance**.  
- The algorithm selects splits that result in the largest reduction in variance.  
- The final prediction at each leaf is the **average of all target values** in that leaf.  

---

### **Key Features of Decision Tree Regressor**  

✅ **Handles Non-Linear Relationships** – Can capture complex patterns in data without requiring feature transformations.  
✅ **Interpretable** – Easy to understand because it follows a tree structure with clear decision rules.  
✅ **Handles Missing Data** – Can split data even when some values are missing.  
✅ **Feature Importance** – Provides insights into which features contribute the most to predictions.  

---

### **Limitations of Decision Tree Regressor**  

❌ **Prone to Overfitting** – A fully grown tree can memorize the training data, leading to poor generalization.  
❌ **Sensitive to Small Changes** – Small variations in data can lead to a completely different tree structure.  
❌ **Not Always the Best for Smooth Predictions** – Since it predicts the **average value per leaf**, it may create step-like (non-smooth) predictions.  

---

### **How to Improve Decision Tree Regression?**  

1. **Pre-Pruning** – Limit tree depth or require a minimum number of samples per leaf to prevent overfitting.  
2. **Post-Pruning** – Grow the tree fully and then remove unnecessary branches.  
3. **Use Ensemble Methods** – Combine multiple decision trees using:  
   - **Random Forest** (multiple trees reduce variance).  
   - **Gradient Boosting** (boosted trees improve accuracy).  

---

### **Example Use Cases**  

🔹 **House Price Prediction** – Predicting property prices based on features like location, size, and number of rooms.  
🔹 **Sales Forecasting** – Estimating future sales based on historical data and market conditions.  
🔹 **Stock Market Trends** – Predicting stock prices based on past performance indicators.

12. What are the advantages and disadvantages of Decision Trees	

Ans. ### **Advantages and Disadvantages of Decision Trees**  

Decision Trees are widely used in machine learning for both **classification** and **regression** tasks due to their simplicity and effectiveness. However, they also have limitations that can impact their performance.  

---

## **✅ Advantages of Decision Trees**  

### **1. Easy to Understand and Interpret**  
- Decision Trees mimic human decision-making, making them highly intuitive.  
- The tree structure provides a **clear set of rules**, making it easy to visualize and explain.  

### **2. Handles Both Numerical and Categorical Data**  
- Unlike models that require feature scaling (e.g., linear regression, SVM), Decision Trees can work with both **continuous** and **categorical** variables without extra preprocessing.  

### **3. Minimal Data Preprocessing Required**  
- No need for **feature scaling** (standardization or normalization).  
- No need for **one-hot encoding** of categorical variables (some implementations handle them directly).  

### **4. Captures Non-Linear Relationships**  
- Unlike linear models, Decision Trees can model **complex and non-linear patterns** in data.  

### **5. Handles Missing Values Well**  
- Decision Trees can handle missing values by **splitting based on available features**, unlike models that require complete data.  

### **6. Provides Feature Importance**  
- Decision Trees can rank **important features**, helping with feature selection and interpretability.  

### **7. Computationally Efficient for Small to Medium Datasets**  
- Building a Decision Tree is relatively fast compared to other machine learning models.  

### **8. Works Well with Large Feature Spaces**  
- Can perform well even when dealing with a high number of features.  

---

## **❌ Disadvantages of Decision Trees**  

### **1. Prone to Overfitting**  
- Fully grown trees can **memorize training data**, leading to poor generalization.  
- **Solution:** Use **pre-pruning (max depth, min samples per split)** or **post-pruning** to remove unnecessary splits.  

### **2. Sensitive to Small Changes in Data**  
- A small change in the training data can lead to a completely different tree structure.  
- **Solution:** Use **ensemble methods** like Random Forest to reduce variance.  

### **3. Biased Toward Dominant Classes**  
- If one class appears more frequently, the tree may become biased toward it.  
- **Solution:** Use **balanced datasets** or apply **class weights**.  

### **4. Poor Performance in Regression (Step-like Predictions)**  
- Decision Trees predict **constant values within leaf nodes**, resulting in **step-wise outputs** rather than smooth predictions.  
- **Solution:** Use **ensemble methods** like Gradient Boosting for better regression results.  

### **5. Computationally Expensive for Deep Trees**  
- Large trees can become slow and memory-intensive, especially for very large datasets.  
- **Solution:** Limit tree depth or use **optimized algorithms** like XGBoost.  

### **6. Cannot Extrapolate Beyond Training Data**  
- Decision Trees predict only within the range of the training data, meaning they do **not generalize well** beyond observed values.  
- **Solution:** Use models like **linear regression** for tasks requiring extrapolation.  

---

## **🔄 How to Overcome These Disadvantages?**

|------------|-------------|
| Overfitting | Pruning, setting max depth, using ensemble methods like Random Forest |
| Instability (sensitive to small changes) | Bagging, Random Forest |
| Biased toward dominant classes | Balance dataset, adjust class weights |
| Poor regression performance | Use ensemble methods like Gradient Boosting |
| High computational cost for deep trees | Limit tree depth, use optimized implementations |

---

### **Final Thoughts**  
Decision Trees are a powerful and interpretable machine learning tool, but they can suffer from **overfitting and instability**. The best way to overcome these weaknesses is by using **pruning** and **ensemble methods** like **Random Forest and Gradient Boosting**.

13.  How does a Decision Tree handle missing values	

Ans. ### **How a Decision Tree Handles Missing Values**  

Decision Trees have built-in mechanisms to deal with missing data efficiently. Unlike models that require complete data (e.g., linear regression), Decision Trees can still make predictions even when some values are missing.  

---

## **Methods for Handling Missing Values in Decision Trees**  

### **1. Splitting by Available Features**  
- During training, a Decision Tree **ignores missing values** and only considers the available data when selecting the best split.  
- If a feature has missing values, the tree evaluates splits based on the **non-missing** data.  

✅ **Advantage:** The model does not discard entire rows just because some values are missing.  

---

### **2. Surrogate Splitting (Advanced Trees like CART)**  
- If a selected feature has missing values, the tree finds a **backup (surrogate) feature** that best mimics the original split.  
- The surrogate feature is then used to direct instances with missing values.  

✅ **Advantage:** Keeps the tree structure intact and ensures that predictions are still possible.  

---

### **3. Assigning the Most Frequent or Average Value**  
- If a feature has missing values, they can be replaced by:  
  - The **most common category** (for categorical data).  
  - The **mean or median** value (for numerical data).  
- This method is often used **before training** if the dataset has a lot of missing values.  

✅ **Advantage:** Simple and easy to implement.  
❌ **Disadvantage:** Can introduce bias, especially if missing values are not random.  

---

### **4. Using Probabilistic Splitting**  
- Instead of sending missing values down a single branch, Decision Trees can **distribute them based on probability**.  
- The data is assigned to branches in proportion to the frequency of non-missing values.  

✅ **Advantage:** Maintains the natural distribution of data.  
❌ **Disadvantage:** Computationally more complex.  

---

### **5. Using "Missing" as a Separate Category**  
- For categorical features, missing values can be treated as a separate category (e.g., “Unknown”).  
- The Decision Tree then learns how to handle these missing values separately.  

✅ **Advantage:** No data is lost.  
❌ **Disadvantage:** Works only for categorical variables, not numerical ones.  

---

## **Which Method Works Best?**  
The best method depends on the dataset:  
- **If missing values are random**, replacing them with the mean/median works well.  
- **If missing values carry meaning**, treating them as a separate category is better.  
- **If missing values are frequent**, using surrogate splits ensures better model accuracy.

14. How does a Decision Tree handle categorical features	

Ans. A **Decision Tree** handles categorical features by splitting the data based on unique values or groups of values in a way that maximizes information gain (or minimizes impurity). Here's how it works:

### **1. Encoding of Categorical Features**
   - If the Decision Tree algorithm supports categorical variables (e.g., **Categorical Decision Trees** in some libraries), it directly splits on category values.
   - If not, categorical features must be **encoded**:
     - **Ordinal Encoding**: Assigns numerical values to categories (useful if the order matters).
     - **One-Hot Encoding**: Converts each category into a separate binary feature (used when categories have no order).

### **2. Splitting on Categorical Variables**
   - For a categorical feature with **few unique values**, the tree can create a separate branch for each value.
   - For a categorical feature with **many unique values**, decision trees might group values based on their impact on the target variable.

### **3. Impurity Calculation**
   - Decision Trees use metrics like **Gini Impurity**, **Entropy (Information Gain)**, or **Chi-Square** to decide the best split.
   - The algorithm tests different splits and selects the one that results in the highest information gain.

### **4. Handling High Cardinality**
   - If a categorical feature has too many unique values (e.g., ZIP codes, product IDs), it can lead to **overfitting**.
   - Possible solutions:
     - **Group similar categories** based on frequency or impact on the target.
     - **Use feature hashing** to reduce dimensionality.
     - **Use tree-based encoding** (e.g., Target Encoding).

### **Example**
If a categorical variable is **"Weather"** with values **{Sunny, Rainy, Cloudy}**, the Decision Tree may split:
- **"Weather = Sunny" → Left branch**
- **"Weather = Rainy or Cloudy" → Right branch** (if they have a similar impact)

### **Key Takeaways**
- Decision Trees can handle categorical data **natively** in some implementations.
- If not, encoding methods (one-hot, ordinal, or target encoding) are required.
- High-cardinality categorical features need special handling to avoid overfitting.

21. * Write a Python program to train a Decision Tree Classifier with a maximum depth of 3 and compare its
accuracy with a fully grown tree

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 

# Load dataset
df = sns.load_dataset('iris')

# Display dataset info
df.head(5)
print(df.info())

# Split dataset
X = df.drop('species', axis=1)
Y = df['species']

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=1)

# Train Decision Tree
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(criterion='entropy')
model.fit(X_train, Y_train)
model1_depth= model.get_depth()
Y_Pred= model.predict(X_test)
from sklearn.metrics import classification_report
print(f'model1 depth is {model1_depth}, and its accuracy is following')
print(classification_report(Y_test, Y_Pred))

model2=DecisionTreeClassifier(criterion='entropy',max_depth=3)
model2.fit(X_train,Y_train)
model2_depth=model2.get_depth()
print(f'model2 depth is {model2_depth}, and its accuracy is following')
Y_pred2=model2.predict(X_test)
print(classification_report(Y_pred2,Y_test))
```

30 Write a Python program to train a Decision Tree Classifier and use GridSearchCV to find the optimal values
for max_depth and min_samples_split.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 

# Load dataset
df = sns.load_dataset('iris')

# Display dataset info
df.head(5)
#print(df.info())

# Split dataset
X = df.drop('species', axis=1)
Y = df['species']
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test= train_test_split(X,Y)
from sklearn.tree import DecisionTreeClassifier
classifier=DecisionTreeClassifier()
params={'max_depth':[3,4,5,6],'min_samples_split':[2,3,4,5,6,7,8]}
from sklearn.model_selection import GridSearchCV

model=GridSearchCV(classifier,param_grid=params,cv=5,scoring='accuracy')
model.fit(X_train,Y_train)
print(model.best_params_)
print(model.best_score_)
```
